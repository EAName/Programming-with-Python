**Revision note (2026):** The heatmaps in this section should not be read as relationships between states or between agencies. Backward difference encoding turns one categorical column into a set of contrast columns (`State_0`, `State_1`, ...) whose values are fixed entirely by which category a row belongs to. The correlations between those columns therefore depend only on the encoding scheme and on how many rows fall in each category; no outcome such as the level of force, armed status or fleeing enters the calculation. High or low values in these heatmaps say nothing about whether a person is likely to experience the same result with one agency or state as with another. Comparing an outcome across states or agencies (for example, the share of deaths by gunshot or of unarmed victims per state, with a chi-square test of independence) would answer that question.
